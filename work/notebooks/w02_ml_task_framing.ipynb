## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary success metric is **Precision@50**.

Precision@50 measures the proportion of the top 50 pages in the ranked review queue that are relevant to the defined review priority.

For this task, pages with better observed search performance, represented by a lower `avg_position`, will be treated as more relevant to the ranking objective.

This metric fits the task because the practical decision is to identify a limited set of pages for human review. A useful ranking approach should place the most relevant pages near the top of the queue.

The relevance definition is based on the observed dataset outcome and is not intended to represent Google's actual ranking algorithm.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could use a single threshold, such as reviewing every page whose average position is worse than a chosen value. However, search performance can depend on multiple observable signals, including impressions, clicks, CTR, engagement, content age, freshness, and word count.

A single fixed rule cannot easily combine these signals or learn patterns between them. An ML ranking approach can learn from historical examples and combine multiple signals to produce a prioritized list of pages.

The ranking output would support a human team in deciding which pages should be reviewed or improved first.

The ML result should be treated as decision-support based on observed data, not as proof of Google's ranking algorithm or as a causal explanation of search rankings.
